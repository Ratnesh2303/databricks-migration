```python
%sql
create catalog scd2
```

```python
%sql
create or replace table scd2.employee_detail.source_employees(
  emp_id int,
  name string,
  dept string,
  position string,
  salary int
)
```

```python
%sql
create or replace table scd2.employee_detail.target_employees(
  emp_id int,
  emp_name string,
  emp_dept string,
  emp_position string,
  emp_salary int,
  start_date date,
  end_date date,
  active boolean
)
```
